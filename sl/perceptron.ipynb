```python
X=df.drop(['placement_status','internet_usage','assignments_completed','previous_score'],axis=1).to_numpy(dtype=np.float32)

encoder=LabelEncoder()
y=encoder.fit_transform(df['placement_status']).astype(np.float32)

X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,stratify=y,test_size=0.2)

y_train = y_train.reshape(-1,1)
```

```python
def sigmoid(z): return 1/(1+np.exp(-z)).astype(np.float32)
```

```python
def perceptron(epoch: int ,X_train: np.ndarray,y_train: np.ndarray,alpha=0.1):
    minloss=float('inf')
    np.random.seed(42)
    w=np.random.random((4,1))
    b=np.random.random()
  
    attributes = (b, w, 0)
    
    for i in range(epoch):
        # 1. Forward pass
        z = X_train.dot(w) + b
        predict = sigmoid(z)
        # backpropogation derivative of loss function and sigmoide is boildowns to pred-actual
        errors =predict-y_train
        w= w-(alpha*((X.T).dot(errors))/len(y_train))

        b=b-(alpha*(np.sum(errors))/len(y_train))

        if i% 100 ==0:
            logloss=np.sum(-(y_train*np.log(predict)+((1-y_train)*np.log(1-predict))))
            
            if logloss<minloss:
                minloss=logloss
                attributes=(b,w,i)
            print(logloss)
    return attributes
```
