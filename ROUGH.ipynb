```python
data = [98, 102, 95, 99, 100, 96, 101, 97, 94, 103, 98, 96, 99, 95, 97]
test_value = 100
```

```python
from scipy import stats
```

```python
p_value= stats.ttest_1samp(a=data,popmean=test_value,alternative='two-sided')
```

```python
alpha=0.05
print(p_value)
```

```python
if p_value<alpha:
    print('reject H0 ' )
else:
    print('Failed to reject')
```
